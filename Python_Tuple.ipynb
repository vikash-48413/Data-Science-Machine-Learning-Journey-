# 📘 Tuple Data Type in Python

---

## ✅ Definition

- A **tuple** is an **ordered**, **immutable** collection of values in Python.
- It is similar to a list, but **cannot be modified** after creation (i.e., no item assignment, deletion, or append).

---

## 🧠 Why Use Tuples?

- ✅ Safer when data must not change.
- ✅ Can be used as dictionary keys (unlike lists).
- ✅ Slightly faster than lists due to immutability.

---

### Update Tuples
-You can convert the tuple into a list, change the list, and convert the list back into a tuple.
- you can treat the tuple like list and apply the list function upon it using this method

````python
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)

print(x)
````

### Unpacking of Tuple

````python
fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)
````

````python
# If the number of variables is less than the number of values, you can add an * to the variable name and the values will be assigned to the variable as a list:
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

(green, yellow, *red) = fruits

print(green)
print(yellow)
print(red)
````

````python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")

(green, *tropic, red) = fruits

print(green)
print(tropic)
print(red)
````

````python
# tuple unpacking
a,b,c = (1,2,3)
print(a,b,c)
````

### Tuple Methods
- count()	Returns the number of times a specified value occurs in a tuple
- index()	Searches the tuple for a specified value and returns the position of where it was found

````python
thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = thistuple.count(5)
print(x)
````

````python
thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = thistuple.index(8)
print(x)
````

### Tuple Function

````python
# 📘 Tuple operations: len, sum, min, max, sorted

# Define a tuple
t = (1, 2, 3, 4)

# Get the length of the tuple
print("Length of tuple:", len(t))

# Get the sum of all elements in the tuple
print("Sum of elements:", sum(t))

# Find the minimum value in the tuple
print("Minimum value:", min(t))

# Find the maximum value in the tuple
print("Maximum value:", max(t))

# Return a sorted list in descending order
print("Sorted in reverse:", sorted(t, reverse=True))
````

### 🔄 Difference Between Lists and Tuples

| Feature               | List                             | Tuple                             |
|------------------------|----------------------------------|------------------------------------|
| **Syntax**             | `[1, 2, 3]`                      | `(1, 2, 3)`                        |
| **Mutability**         | ✅ Mutable                       | ❌ Immutable                       |
| **Speed**              | Slower                          | Faster                            |
| **Memory**             | Consumes more memory            | Consumes less memory              |
| **Built-in Functions** | Rich set of methods (e.g., `append`, `remove`) | Limited methods                  |
| **Error Prone**        | More error-prone due to changes | Less error-prone (can’t be modified) |
| **Usability**          | Suitable for dynamic data       | Suitable for fixed/constant data  |


````python
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*10
print('List time',time.time()-start)

start = time.time()
for i in T:
  i*10
print('Tuple time',time.time()-start)
````

````python
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
````

````python
t = (10, 20, 30, 40, 50)

# Using `in`
print(30 in t)       # ✅ True (30 is in the tuple)
print(100 in t)      # ❌ False (100 is not in the tuple)

# Using `not in`
print(100 not in t)  # ✅ True (100 is not in the tuple)
print(10 not in t)   # ❌ False (10 is in the tuple)
````

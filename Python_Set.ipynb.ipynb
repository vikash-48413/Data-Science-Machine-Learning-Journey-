# 🧠 Python Sets - Quick Notes

## 🔹 What is a Set?
A **set** is a built-in Python collection that:
- Is **unordered**
- Is **mutable** (can add/remove items)
- Contains **unique, immutable elements** (no duplicates)

---

## 📌 Key Characteristics of Sets
| Feature           | Description                                                                 |
|------------------|-----------------------------------------------------------------------------|
| **Unordered**     | No guarantee of order — elements have no index                             |
| **Mutable**       | Elements can be added/removed                                               |
| **No Duplicates** | Duplicate values are automatically removed                                 |
| **Hashable Items Only** | Set elements must be immutable types like int, str, tuple (not list, dict) |
| **True == 1**, **False == 0** | Considered duplicates: `{True, 1}` ➝ `{True}` and `{0, False}` ➝ `{False}` |

---

### Set Operation

````python
# Using |  operator -> It only works with the set
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1 | set2 | set3 |set4
print(myset)
````

````python
# With multiple sets - > we can use the .union() method for this
set1 = {"a", "b", "c"}
l2 = [1, 2, 3]
set3 = {"John", "Elena"}
set4 = set("python")

myset = set1.union(l2, set3, set4)
print(myset)
````

- Note: The  | operator only allows you to join sets with sets, and not with other data types like you can with the  union() method.

### Frozenset
Frozen set is just an immutable version of a Python set object
- frozenset is ideal when you really need fast membership plus hashability.
- created by the built-in frozenset() constructor
- Unordered, stores only unique items, and each item must be hashable (immutable).
- Because it’s immutable, it is hashable itself and can be used as a dictionary key or as an element of another set.


````python
empty = frozenset()                     # canonical empty object
f1    = frozenset([1, 2, 3])            # from any iterable
f2    = frozenset({x**2 for x in range(4)})   # from comprehension
f1,f2
````

````python
# Set algebra (returns frozenset)
a, b = frozenset({1, 2}), frozenset({2, 3})
u = a | b         # union
i = a & b         # intersection
d = a - b         # difference
x = a ^ b         # symmetric difference
u,i,d,x
````

````python
# No mutation
a.add(4)          # AttributeError
````

````python
# When to use
# 2D sets
fs = frozenset([1,2,frozenset([3,4])])
fs
````

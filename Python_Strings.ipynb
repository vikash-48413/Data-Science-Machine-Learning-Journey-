````python
# Using f-strings

# Syntax: f"literal text {expression} literal text {expression}..."
# Rule: The string literal must be prefixed with f or F.
# Expressions inside the curly braces {} are evaluated and their results are converted to strings.

# Very readable and often the preferred method for embedding variables or expressions into strings.
# Efficient as they are processed at runtime into a single string.
# Automatically handles the conversion of non-string types within the curly braces to strings.
# Example:

age = 20

txt = f"My name is Vikash, I am {age}"
print(txt)
````

````python
# Using Adjacent String Literals -

# Adjacent string literals are automatically concatenated by the Python parser.
# Syntax: "string1" "string2"
# Rule: This only works for string literals, not variables.
# Details: Useful for breaking long strings across multiple lines in source code.

long_string = ("This is the first part. "
                       "This is the second part. "
                       "This is the third part.")
print(long_string)
````

````python
# Write a program to count the number of words in a string without split()
s = input("Enter the text:")
l = []
temp = ''
for i in s:
  if i != ' ':
    temp = temp + i
  else:
    l.append(temp)
    temp = ''
l.append(temp)
print(l)
````

````python
s = input("Enter the text: ")
words = []
temp = ''

for char in s:
    if char != ' ':
        temp += char
    elif temp:
        words.append(temp)
        temp = ''

# Add the last word if any
if temp:
    words.append(temp)

print("Words:", words)
print("Number of words:", len(words))
````

````python
# Write a python program to convert a string to title case without using the title()
s = input('enter the string')

L = []
for i in s.split():
  L.append(i[0].upper() + i[1:].lower())

print(" ".join(L))
````

````python
# Write a program that can convert an integer to string.

number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result # this line will return the remainder and concatenate the result form left to right
  number = number//10

print(result)
print(type(result))
````

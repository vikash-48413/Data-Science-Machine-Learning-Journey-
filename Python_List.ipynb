# 📘 Python List
## ✅ Definition:
A **list** in Python is a **built-in data type** used to **store multiple items** in a **single variable**. It is an **ordered**, **mutable**, and **indexable** collection that can hold elements of **different data types**, including other lists.

---

## 🧠 Key Features:
- ✅ **Ordered**: Elements maintain their insertion order.
- ✅ **Mutable**: You can add, remove, or change items.
- ✅ **Indexable**: Access items using their position (starting from 0).
- ✅ **Allows duplicates**: Repeated values are allowed.
- ✅ **Heterogeneous**: Can contain different data types (int, str, float, etc.).
- ✅ **Dynamic Size**: Lists can grow or shrink during runtime.


<small>

## Remove List Items

````python
# pop() -->  removes the specified index
l = [1,2,True,4,5]
l.pop(2)
l

# Note : If you do not specify the index, the pop() method removes the last item.
````

````python
# remove() --> removes the specified item.
l = [1,2,True,4,5]
l.remove(2)
l

# Note : If there are more than one item with the specified value, the remove() method removes the first occurrence.
````

````python
# del keyword --> The del keyword also removes the specified index:
l = [1,2,3,True,5]
del l[3]
l
````

````python
# del keyword --> The del keyword is also used to delete the whole list:
l = [1,2,3,True,5]
del l
l
````

````python
l = [1, 2, 3, 2, 4, 2, 5]
item_to_remove = 2

# Create a new list excluding the item to remove
new_list = [item for item in l if item != item_to_remove]

print(new_list)
````

````python
# Clear --> The clear() method empties the list.
# The list still remains, but it has no content.
l = [1,2,3,4,5]
l.clear()
l
````

#Python - List Comprehension
-List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

- SYNTAX : newlist = [expression for item in iterable if condition == True]

- used to avoid the for loop
- The return value is a new list, leaving the old list unchanged
- The condition is like a filter that only accepts the items that evaluate to True.


-Advantages of List Comprehension
- More time-efficient and space-efficient than loops.
- Require fewer lines of code.
- Transforms iterative statement into a formula.


````python
# Without list comprehension
l = [1,2,3,4,5,6,7,8]
new_list = []

for i in l :
  if i>3:
    new_list.append(i)

print(new_list)
````

````python
# With list comprehension
[print(i) for i in l if i > 3]
````

````python
[i for i in l if i > 3]
````

````python
newlist = [x for x in range(10) if x < 5]
print(newlist)
````

````python
newlist = ['hello' for x in l]
newlist
````

````python
newlist = [x if x != 8 else "orange" for x in l]
newlist
````

````python
L = [i for i in range(1,11)]
print(L)
````

````python
# Scalar Multiplication on a vector
s = [1,2,3,4,5]
v = -3
[v*i for i in s]
````

````python
# Add squares
L = [1,2,3,4,5]

[i**2 for i in L]
````

````python
# Print all numbers divisible by 5 in the range of 1 to 50
[i for i in range(1,51) if i%5==0]
````

````python
# find languages which start with letter p
languages = ['java','python','php','c','javascript']
[l for l in languages if l.startswith('p')]
````

````python
# Print a (3,3) matrix using list comprehension -> Nested List comprehension
[[i*j for i in range(1,4)] for j in range(1,4)]
````

````python
# cartesian products -> List comprehension on 2 lists together
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i*j for i in L1 for j in L2]
````

## Python - Sort Lists
List objects have a sort() method that will sort the list alphanumerically, ascending, by default:



````python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print(thislist)
````

````python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse = True)
print(thislist)
````

````python
# Customize Sort Function
# - > Sort the list based on how close the number is to 50:
def myfunc(n):
  return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key = myfunc)
print(thislist)
````

````python
# Case Insensitive Sort
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)
````

````python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(thislist)
````

````python
# Reverse list
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.reverse()
print(thislist)
````

## Copy
You cannot copy a list simply by typing list2 = list1, because: list2 will only be a reference to list1, and changes made in list1 will automatically also be made in list2.

````python
# copy()
thislist = ["apple", "banana", "cherry"]
mylist = thislist.copy()
print(mylist)
````

````python
# list()
thislist = ["apple", "banana", "cherry"]
mylist = list(thislist)
print(mylist)
````

````python
# Use the slice Operator
thislist = ["apple", "banana", "cherry"]
mylist = thislist[:]
print(mylist)
````

#List Methods
Python has a set of built-in methods that you can use on lists.

Method	Description
- append()	Adds an element at the end of the list
- clear()	Removes all the elements from the list
- copy()	Returns a copy of the list
- count()	Returns the number of elements with the specified value
- extend()	Add the elements of a list (or any iterable), to the end of the current list
- index()	Returns the index of the first element with the specified value
- insert()	Adds an element at the specified position
- pop()	Removes the element at the specified position
- remove()	Removes the item with the specified value
- reverse()	Reverses the order of the list
- sort()	Sorts the list

````python
# Write a program to merge 2 list without using the + operator
L1 = [1, 2, 3, 4]
L2 = [5, 6, 7, 8]

merged = [x for group in (L1, L2) for x in group]
print(merged)
````

````python
# Write a program to replace an item with a different item if found in the list
L = [1, 2, 3, 4, 5, 3]

# Replace 3 with 300
L = [300 if i == 3 else i for i in L]

print(L)
````

````python
# Write a program that can convert a 2D list to 1D list
L = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [item for sublist in L for item in sublist]
print(flattened)
````

````python
# Write a program to remove duplicate items from a list
L = [1, 2, 1, 2, 3, 4, 5, 3, 4]

L_no_duplicates = list(set(L))
print(L_no_duplicates)
````

````python
# Write a program to check if a list is in ascending order or not
L = [1, 2, 3, 4, 5]

if all(L[i] <= L[i + 1] for i in range(len(L) - 1)):
    print("The list is in ascending order.")
else:
    print("The list is NOT in ascending order.")
````

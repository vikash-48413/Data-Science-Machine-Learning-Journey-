# ✅ Python Conditionals

```
- Python supports the following logical conditions:
  - Equals:               a == b
  - Not Equals:           a != b
  - Less than:            a < b
  - Less than or equal:   a <= b
  - Greater than:         a > b
  - Greater than or equal:a >= b

Notes:
  - if keyword is use for conditionals
  - indentation is used to define the scope and in conditional it is used heavily
  - Python doesn’t "look ahead" — it stops checking once a match is found from if, elif and else




if condition1:
    # run this
elif condition2:
    # run this
else:
    # nothing matched, run fallback

````python
# if
a = 10
b = 20
if a < b:
  print("b is grater than a")
````

````python
# Without Identation
a = 33
b = 200
if b > a:
print("b is greater than a") # you will get an error
````

````python
# Elif - The elif keyword is Python's way of saying "if the previous conditions were not true, then try this condition".
# It is optional
a = 33
b = 33
if a > b:
  print("a is grater than b")
elif a == b:
  print("a is equal to b")
````

````python
# Else - The else keyword catches anything which isn't caught by the preceding conditions.
menu = input("What would you like to order pizza/burger? ").lower().replace(" ", "")

if menu == "pizza":
    print("Serving pizza")
elif menu == "burger":
    print("Serving burger")
else:
    print("Serving Water")
````

````python
# Short Hand If
if 5 > 3:print("This is shorthand if")
````

````python
# Short Hand if...else -> <true_expression> if <condition> else <false_expression>
# This technique is known as Ternary Operators, or Conditional Expressions.

print("5") if 5 > 3 else print("3")
````

````python
# Short Hand if...else with 3 condition
a = 5
b = 3
print("A") if a > b else print("equal") if a == b else print("B")
````

````python
# and - > The and keyword is a logical operator, and is used to combine conditional statements:
if 5 > 4 and 4 < 5: # Here the both condition should be true
  print("We got the number")
````

````python
# or - same as and to combine the statements but true if only one condition is true
if 5 > 4 or 4 < 5:
  print("We got the number")
````

````python
# not - The not keyword is a logical operator, and is used to reverse the result of the conditional statement:
a = 10
b = 20
if not a > b:
  print("a is NOT greater than b")
````

````python
a = 33
b = 200
if not a > b:
  print("True")
````

````python
# Nested If
age = int(input("Enter your age:") or 0) # Here 0 is default input age if user skips the input

if age >=0 and age <= 100:
  if age >=18:
    if age >65:
      print("Senior Citizen")
    else:
      print("Adult")
  else:
    if age >= 13:
      print('Teenager')
    else:
      print("Child")
else:
  print("Please enter the valid age")
````

#### Match statement


```
match subject:
    case pattern1:
        action1
    case pattern2:
        action2
    case _:
        fallback_action
```



````python
# Weekday using Match statement
day = 3

match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
  case _:
    print("Invalid Day")
````

````python
# Weekend for Us
day = 6
match day:
  case 1 | 2 | 3 | 4 | 5:
    print("Today is a weekday")
  case 6 | 7:
    print("I love weekends!")
````

````python
month = 5
day = 4
match day:
  case 1 | 2 | 3 | 4 | 5 if month == 4:
    print("A weekday in April")
  case 1 | 2 | 3 | 4 | 5 if month == 5:
    print("A weekday in May")
  case _:
    print("No match")
````

#### while loop
- With the while loop we can execute a set of statements as long as a condition is true.
- remember to increment i, or else the loop will continue forever.


````python
i = 1
while i < 5:
  print(i)
  i += 1
````

````python
# break statemnt -> With the break statement we can stop the loop even if the while condition is true:
i = 0
while i <= 10:
  print(i)
  if i == 5:
    break
  i += 1
````

````python
# Continue = With the continue statement we can stop the current iteration, and continue with the next:
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
````

````python
# While loop with else:
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
````

````python
# Printing Table using while loop
number = int(input('enter the number') or 2)

i = 1

while i<11:
  print(number,'*',i,'=',number * i)
  i += 1
````

#### For loop
- A for loop in Python is used to iterate over a sequence (like a list, tuple, string, or range) and execute a block of code multiple times, once for each item in the sequence.


```
syntax :
for variable in sequence:
    # do something

```



````python
fruits = ['apple', 'banana', 'mango']
for fruit in fruits:
    print(fruit)
````

````python
# Using range()
for  i in range (5):
    print(i)
````

````python
# Step
for i in range(1, 10, 2):
    print(i)
````

````python
# Looping through character
for i in "Vikash":
  print(i)
````

````python
# With else statement
for i in range(3):
    print(i)
else:
    print("Loop finished!")
````

````python
# Using break statement
for i in range(5):
    if i == 3:
        break
    print(i)
````

````python
# using continue statement
for i in range(5):
  if i == 3:
    continue
  print(i)
````

````python
# Star printing using for loop
for i in range(1, 6):
    print("⭐" * i)
````

````python
# Nested for loop
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
````

````python
# Using pass statement
for x in [0, 1, 2]:
  pass
````

````python
# Traffic Light simulator using the for loop
colors = ["red", "yellow", "green"]
for color in colors:
    if color == "red":
      print("Stop🛑")
    elif color == 'yellow':
      print("Wait")
    else:
      print("Go")
````

````python
for idx, val in enumerate(['a', 'b', 'c']):
    print(idx, val)
````

````python
# Find the max value in a list without using max().
numbers = [213, 23422, 12, 99, 45]
max_val = numbers[0]
for num in numbers[1:]:
    if num > max_val:
        max_val = num
print("Max value is:", max_val)
````

````python
# Loop through multiple lists with zip()
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"{name} scored {score}")
````

````python
#  Nested loops for pattern printing (stars)
for i in range(1,6):
  for j in range(i):
    print("*",end ="") # This keeps the for loop in the same line until the complete for loop execution
  print()
````

````python
rows = 5
for i in range(1, rows + 1):
    print(' ' * (rows - i) + '* ' * i)
````
